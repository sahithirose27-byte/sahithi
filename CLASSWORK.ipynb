```python
class student:
  clg="ABC College"# class variable

  def __init__(self,name):
    self.name=name

  def show_name(self):# instance method
    print(self.name)

  @classmethod
  def show_clg(cls):# class method
    print(cls.clg)

  @staticmethod
  def add(a,b):#static method
    return a+b

s=student("rahul")
s.show_name()
student.show_clg()
print(student.add(2,3))
```

```python
# abstraction
from abc import ABC,abstractmethod

class payment(ABC):
  @abstractmethod
  def pay(self,amount):
    pass

class CCP(payment):
  def pay(self,amount):
    print("paid",amount,"using Credit card")

class UPI(payment):
  def pay(self,amount):
    print("paid",amount,"using UPI")

class Cash(payment):
  def pay(self,amount):
    print("paid",amount,"using Cash")

payment=Cash()
payment.pay(100)
```

```python
# abstraction
from abc import ABC,abstractmethod

class payment(ABC):
  @abstractmethod
  def pay(self,amount):
    pass

class CCP(payment):
  def pay(self,amount):
    print("paid",amount,"using Credit card")

class UPI(payment):
  def pay(self,amount):
    print("paid",amount,"using UPI")

CCP().pay(5000)
UPI().pay(1000)
```

```python
# encapsulation
class BankAccount:
  def __init__(self,balance):
    self.__balance=balance

  def deposit(self,amount):
    if amount>0:
      self.__balance+=amount

  def show_balance(self):
    print(self.__balance)

account=BankAccount(5000)
account.deposit(2000)
account.show_balance()
```

```python
class BankAccount:
  def __init__(self,total):
    self.__total=total

  def deposit(self,amount):
    if amount>0:
      self.__total+=amount
      print(amount,"is added to actual amount")

  def withdraw(self,amount):
    if amount>0:
      self.__total-=amount
      print(amount,"is withdrawn from total amount")

  def fetch_balance(self):
    print("balnce:",self.__total)

account=BankAccount(10000)
account.deposit(2000)
account.withdraw(1000)
account.fetch_balance()
```

```python
# Polymorphism
#run time(method overriding)
#compile time(method overloading)

#method overloading--> same method name with diff. parameters

class operations:
  def add(self,a,b=None,c=None,d=None):
    print(a,b,c,d)
obj=operations()
obj.add(1,2,3,5)
```

```python
#method overriding--> same method name with same parameters
class operations:
  def add(self,a):
    print("1st function")
  def add(self,a):
    print("2nd function")
  def add(self,a):
    print("3rd function")
obj=operations()
obj.add(5)
```

```python
class addition:
  def add(self,a,b):
    print(a+b)
class subtraction(addition):
  def sub(self,a,b):
    print(a-b)
class multiplication(subtraction):
  def mult(self,a,b):
    print(a*b)
obj=multiplication()
obj.add(10,20)
obj.sub(10,50)
obj.mult(10,3)
```

```python
class addition:
  def add(self):
    print("adding")
class subtraction(addition):
  def sub(self):
    print("subtracting")
class multiplication(subtraction):
  def mult(self):
    print("multiplication")
obj=multiplication()
obj.add()
obj.sub()
obj.mult()
```
