```python
#6 Employee Management Using Inheritance
class Employee:
  def __init__(self,emp_name,emp_ID,salary):
    self.emp_name=emp_name
    self.emp_ID=emp_ID
    self.salary=salary
  def display(self):
    print("employee name:",self.emp_name)
    print("employee ID:",self.emp_ID)
    print("salary:",self.salary)

class Developer(Employee):
  def __init__(self,emp_name,emp_ID,salary,programming_language):
    super().__init__(emp_name,emp_ID,salary)
    self.programming_language=programming_language
  def display_details(self):
    print("\n  DEVELOPER DETAILS")
    print("employee name:",self.emp_name)
    print("employee ID:",self.emp_ID)
    print("salary:",self.salary)
    print("programming language:",self.programming_language)

class Manager(Employee):
  def __init__(self,emp_name,emp_ID,salary,team_size):
    super().__init__(emp_name,emp_ID,salary)
    self.team_size=team_size
  def display_details(self):
    print("\n  MANAGER DETAILS")
    print("employee name:",self.emp_name)
    print("employee ID:",self.emp_ID)
    print("salary:",self.salary)
    print("team size:",self.team_size)

dev1=Developer("Alice",101,50000,"Python")
dev2=Developer("Bob",102,60000,"Java")

mgr1=Manager("Charlie",103,80000,5)
mgr2=Manager("David",104,90000,3)

dev1.display_details()
dev2.display_details()

mgr1.display_details()
mgr2.display_details()
```

```python
#9 Abstract Class -- Shape Calculator
from abc import ABC,abstractmethod
class shape(ABC):
  @abstractmethod
  def calculate_area(self):
    pass

class circle(shape):
  def __init__(self,radius):
    self.radius=radius
  def calculate_area(self):
    return 3.14*self.radius*self.radius

class rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def calculate_area(self):
    return self.length*self.width

class triangle(shape):
  def __init__(self,base,height):
    self.base=base
    self.height=height
  def calculate_area(self):
    return 0.5*self.base*self.height

Circle=circle(5)
Rectangle=rectangle(4,6)
Triangle=triangle(3,4)

print("circle area:",Circle.calculate_area())
print("rectangle area:",Rectangle.calculate_area())
print("triangle area:",Triangle.calculate_area())
```

```python
#10 Mini Project -- Library Management System
class Book:
  def __init__(self,book_id,name,author):
    self.book_id=book_id
    self.name=name
    self.author=author
    self.is_available=True
  def __str__(self):
    status="available" if self.is_available else "not available"
    return f"{self.name} by {self.author} (ID: {self.book_id}) - {status}"

class Library:
  def __init__(self):
    self.books=[] # self.books is a list of Book objects

  def add_book(self,book_id,name,author):
    # Check if book_id already exists in the list
    for book in self.books:
      if book.book_id == book_id:
        print(f"Book with ID {book_id} already exists.")
        return

    new_book=Book(book_id,name,author)
    self.books.append(new_book)
    print("Book added successfully.")

  def display_books(self):
    if not self.books:
      print("List of books in library is empty")
      return
    print("\n--- Library Catalog ---")
    for book in self.books: # Iterate directly over the list
      print(book)

  def search_book(self,query):
    query_str = str(query).strip().lower()
    results=[]

    for book in self.books: # Iterate directly over the list
      # Check if book_id matches or query is in book name (case-insensitive)
      if str(book.book_id).lower() == query_str or query_str in book.name.lower():
        results.append(book)

    if not results:
      print("No books found")
      return
    print("Search results:")
    for book in results:
        print(book)
    return results

  def _find_book_by_id(self, book_id):
    """Helper method to find a book object by its ID from the list."""
    for book in self.books:
      if book.book_id == book_id:
        return book
    return None

  def issue_book(self,book_id):
    book = self._find_book_by_id(book_id) # Use helper to find book
    if not book:
      print(f"Book with ID {book_id} not found.")
      return
    if not book.is_available:
      print(f"Book '{book.name}' (ID: {book_id}) is not available or is issued to someone else.")
      return
    book.is_available=False
    print(f"Book '{book.name}' (ID: {book_id}) issued successfully.")

  def return_book(self,book_id):
    book = self._find_book_by_id(book_id) # Use helper to find book
    if not book:
      print(f"Book w2ith ID {book_id} not found.")
      return
    if book.is_available:
      print(f"Book '{book.name}' (ID: {book_id}) is already available.")
      return
    book.is_available=True
    print(f"Book '{book.name}' (ID: {book_id}) returned successfully.")

def main():
  library=Library()
  library.add_book(101,"The Great Gatsby","F. Scott Fitzgerald")
  library.add_book(202,"To Kill a Mockingbird","Harper Lee")
  library.add_book(303,"1984","George Orwell")

  while True:
    print("\n--- Library Management System ---")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Books")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Exit")

    choice=input("\nEnter your choice (1-6):").strip()
    if choice =='1':
      try:
        b_id=int(input("Enter book ID:").strip())
        name=input("Enter book name:").strip()
        author=input("Enter book author:").strip()
        if name and author:
          library.add_book(b_id,name,author)
        else:
          print("Invalid input: Name and Author cannot be empty.")
      except ValueError:
        print("Invalid input: Book ID must be an integer.")
    elif choice=='2':
      library.display_books()
    elif choice=='3':
      query=input("Enter search query (book ID or name):").strip()
      if query:
        # If query is purely digits, attempt to search by ID (as int)
        if query.isdigit():
            library.search_book(int(query))
        else: # Otherwise, search by name (as string)
            library.search_book(query)
      else:
        print("Invalid input: Search query cannot be empty.")
    elif choice=='4':
      try:
        book_id=int(input("Enter book ID to issue:").strip())
        library.issue_book(book_id)
      except ValueError:
        print("Invalid input: Book ID must be an integer.")
    elif choice=='5':
      try:
        book_id=int(input("Enter book ID to return:").strip())
        library.return_book(book_id)
      except ValueError:
        print("Invalid input: Book ID must be an integer.")
    elif choice=='6':
      print("Exiting Library Management System. Goodbye!")
      break
    else:
      print("Invalid choice. Please enter a number between 1 and 6.")

main()
```
